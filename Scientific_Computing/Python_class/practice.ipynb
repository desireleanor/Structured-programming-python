```python
x = int(input("Enter marks of Math: "))
y = int(input("Enter marks of Science: "))
z = int(input("Enter marks of Geography: "))
a = int(input("Enter marks of History: "))
b = int(input("Enter marks of English: "))
avg = (x+y+z+a+b)/5
print("Average of the five subject marks is: ", avg)
if avg >= 80:
    print("You are an outstanding student.")
if 80>avg>=70:
    print("You are a good student.")
if 70>avg>=60:
    print("You are an average student.")
if 60>avg>=50:
    print("You are a below-average student.")
if 50>avg>=40:
    print("You are a poor student.")
if avg<40:
    print("You need extraordinary efforts.")
```

```python
import math
x = int(input("Enter the first length of the triangle: "))
y = int(input("Enter the second length of the triangle: "))
z = int(input("Enter the third length of the triangle: "))

if x + y > z and x + z > y and y + z > x:   #IMPORTANT
    s = (x + y + z) / 2
    area = math.sqrt(s*(s-x)*(s-y)*(s-z))    #heron's formula
    print("Area =", area)
else:
    print("Triangle is invalid")
```
